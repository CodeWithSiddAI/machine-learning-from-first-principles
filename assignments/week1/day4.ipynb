### Task 2 (Bonus)

Modify your function so you can approximate the derivative of **any** single-variable function passed to it.

```python
import numpy as np
```

```python
def f(x):
    return x**3

def numerical_derivative(f, x, h=1e-5):
    return (f(x+h) - f(x)) / h 

def analytical_derivative(x):
    return 3 * (x**2)

print(f"{'x':<5}{'Numerical':<15}{'Analytical':<15}{'Error':<15}")
print("-" * 50)

for x in [1, 2, 5]:
    numerical = numerical_derivative(f, x)
    analytical = analytical_derivative(x)
    error = abs(numerical - analytical)

    print(f"{x:<5}{numerical:<15.6f}{analytical:<15.6f}{error:<15.6f}")
```

```python
# we can also write it using pandas style
import pandas as pd

df = pd.DataFrame({
    "x": [1, 2, 5],
    "Numerical": [3.00003, 12.00006, 75.00015],
    "Analytical": [3, 12, 75]
})

df["Error"] = abs(df["Numerical"] - df["Analytical"])

print(df)
```
